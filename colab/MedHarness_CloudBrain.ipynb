# MedHarness — Cloud Brain (free T4)

Spins up the dual-model clinical engine (**Med-R1-8B** + **MedGemma-4B**, both
GPTQ W4A16 Marlin, hot in VRAM) behind a bearer-authenticated Cloudflare
tunnel.

**How to use**
1. **Runtime → Change runtime type → T4 GPU**, then run the two cells below.
2. Copy the **token** printed in step 1 into the MedHarness desktop app
   (Settings → Cloud Brain → Session token).
3. When the bootstrap banner prints the `https://….trycloudflare.com` URL,
   paste it into the app → **Connect**.

> Bootstrap is fetched live from `https://github.com/KeshavCracks/openagent` so you always run the audited latest version.


```python
# 1) Parameters — leave the token blank to auto-generate one
SESSION_BEARER_TOKEN = ""   # paste the token from your MedHarness app, or leave blank
MEDR1_GPTQ_REPO = ""        # optional: GPTQ W4A16 checkpoint (HF id or local dir)
MEDGEMMA_GPTQ_REPO = ""     # optional: GPTQ W4A16 checkpoint (HF id or local dir)

import secrets
if not SESSION_BEARER_TOKEN:
    SESSION_BEARER_TOKEN = secrets.token_urlsafe(32)
print("Session token (paste into the MedHarness app → Settings):")
print(" ", SESSION_BEARER_TOKEN)
```

```python
# 2) Fetch the audited bootstrap from GitHub and deploy (~3–8 min)
import os
os.environ["SESSION_BEARER_TOKEN"] = SESSION_BEARER_TOKEN
if MEDR1_GPTQ_REPO:
    os.environ["MEDR1_GPTQ_REPO"] = MEDR1_GPTQ_REPO
if MEDGEMMA_GPTQ_REPO:
    os.environ["MEDGEMMA_GPTQ_REPO"] = MEDGEMMA_GPTQ_REPO

!curl -fsSL "https://raw.githubusercontent.com/KeshavCracks/openagent/main/core/cloud_bootstrap.sh" -o /content/medharness_bootstrap.sh
!bash /content/medharness_bootstrap.sh

# When the CLOUD BRAIN — ONLINE banner appears, copy the tunnel URL into the app.
```
